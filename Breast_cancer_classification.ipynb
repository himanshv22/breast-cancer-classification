```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load the Breast Cancer dataset
data = load_breast_cancer()
x= data.data
y = data.target

# Split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Initialize the classifiers
rf = RandomForestClassifier(n_estimators=100, random_state=42)
dt = DecisionTreeClassifier(random_state=42)
ab = AdaBoostClassifier(n_estimators=50, random_state=42)

# Train the classifiers
rf.fit(x_train, y_train)
dt.fit(x_train, y_train)
ab.fit(x_train, y_train)

# Predict the labels for the test set
rf_pred = rf.predict(x_test)
dt_pred = dt.predict(x_test)
ab_pred = ab.predict(x_test)

# Calculate accuracy for each model
rf_acc = accuracy_score(y_test, rf_pred)
dt_acc = accuracy_score(y_test, dt_pred)
ab_acc = accuracy_score(y_test, ab_pred)

# Create a bar graph to visualize accuracies
models = ['Random Forest', 'Decision Tree', 'AdaBoost']
accuracies = [rf_acc, dt_acc, ab_acc]

plt.figure(figsize=(10, 6))
plt.bar(models, accuracies, color=['skyblue', 'lightcoral', 'lightgreen'])
plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.title('Classifier Accuracies on Breast Cancer Dataset')
plt.ylim(0.85, 1.0) # Set y-axis limits for better visualization of differences
for i, acc in enumerate(accuracies):
    plt.text(i, acc + 0.01, f'{acc:.2f}', ha='center') # Add accuracy values on top of bars
plt.show()
```
